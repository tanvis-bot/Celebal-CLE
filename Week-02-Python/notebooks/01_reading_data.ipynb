```python
import pandas as pd

file_path = "../data/raw/yellow_tripdata_sample.csv"

df = pd.read_csv(file_path)

print("CSV loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
cleaned_df = df.copy()

print("Working copy created!")
print("Shape:", cleaned_df.shape)
```

```python
print("Missing values before cleaning:")
print(cleaned_df.isnull().sum().sum())

cleaned_df = cleaned_df.dropna()

print("Missing values after cleaning:")
print(cleaned_df.isnull().sum().sum())
```

```python
cleaned_df["tpep_pickup_datetime"] = pd.to_datetime(
    cleaned_df["tpep_pickup_datetime"]
)

cleaned_df["tpep_dropoff_datetime"] = pd.to_datetime(
    cleaned_df["tpep_dropoff_datetime"]
)

print(cleaned_df[
    ["tpep_pickup_datetime", "tpep_dropoff_datetime"]
].dtypes)
```

```python
print("Duplicates before cleaning:", cleaned_df.duplicated().sum())

cleaned_df = cleaned_df.drop_duplicates()

print("Duplicates after cleaning:", cleaned_df.duplicated().sum())

print("Final shape:", cleaned_df.shape)
```

```python
output_path = "../data/processed/yellow_tripdata_cleaned.parquet"

cleaned_df.to_parquet(output_path, index=False)

print("Cleaned dataset saved successfully!")
```

```python
check_df = pd.read_parquet(output_path)

print("Saved dataset shape:", check_df.shape)
print("Saved dataset columns:", len(check_df.columns))
```

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logger = logging.getLogger(__name__)

logger.info("Logging configured successfully!")
```

```python
import requests
import json

def fetch_api_data(url, output_path):

    try:
        logger.info("Fetching data from API...")

        response = requests.get(url, timeout=10)

        response.raise_for_status()

        data = response.json()

        with open(output_path, "w") as file:
            json.dump(data, file, indent=4)

        logger.info("Data fetched and saved successfully!")

        return data

    except requests.exceptions.RequestException as error:
        logger.error(f"API request failed: {error}")

    except (ValueError, json.JSONDecodeError) as error:
        logger.error(f"Invalid JSON response: {error}")

    except OSError as error:
        logger.error(f"File operation failed: {error}")

    return None
```

```python
api_url = "https://jsonplaceholder.typicode.com/posts"
output_file = "../data/raw/api_posts_function.json"

result = fetch_api_data(api_url, output_file)

if result is not None:
    print("Total posts fetched:", len(result))
```

```python
invalid_url = "https://jsonplaceholder.typicode.com/invalid"

test_result = fetch_api_data(
    invalid_url,
    "../data/raw/test.json"
)

print("Function result:", test_result)
```
